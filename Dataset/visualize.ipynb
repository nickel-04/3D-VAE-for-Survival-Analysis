```python
import random
import matplotlib.pyplot as plt
import torch
import torch
import matplotlib.pyplot as plt
import random
import pandas as pd
import sys
import os
sys.path.append(os.path.curdir + '/..')
from Dataset.custom_dataset import get_data_loaders
from pycox.evaluation import EvalSurv
from statsmodels.stats.proportion import proportion_confint
import Dataset
print(Dataset.__name__)

def visualize_random_slices_before_after(dataloader, num_images=3, slice_idx=None):
    """
    Visualize random slices from images before and after transformations.

    Args:
        dataloader (DataLoader): DataLoader to sample images from.
        num_images (int): Number of images to sample and display.
        slice_idx (int): Index of the slice to view. If None, a random slice is selected.
    """
    # Set up for the number of images to visualize
    displayed_images = 0

    for idx in range(len(dataloader.dataset)):
        if displayed_images >= num_images:
            break  # Stop after visualizing the desired number of images

        # Get the patient folder and original DICOM files
        patient_folder = dataloader.dataset.patient_folders[idx]
        image_path = load_images(patient_folder)  # Original NIfTI path (before transformation)

        # Load original NIfTI data before transformations
        if image_path is not None:
            original_image = nib.load(image_path).get_fdata()
            depth = original_image.shape[2]

            # Pick a random slice if not provided
            if slice_idx is None:
                selected_slice = random.randint(0, depth - 1)
            else:
                selected_slice = slice_idx

            # Fetch the transformed image from the DataLoader
            transformed_image_tensor, _, _ = dataloader.dataset[idx]

            # Check if the transformed image is a valid tensor
            if transformed_image_tensor is None or not isinstance(transformed_image_tensor, torch.Tensor):
                print(f"Skipping patient at index {idx} due to invalid transformed image.")
                continue

            # Plotting: before and after transformation
            fig, axs = plt.subplots(1, 2, figsize=(12, 6))
            fig.suptitle(f"Patient ID: {os.path.basename(patient_folder)} | Slice: {selected_slice}")

            # Display original (before transformation)
            axs[0].imshow(original_image[:, :, selected_slice], cmap="gray")
            axs[0].set_title("Original Image (Before Transform)")
            axs[0].axis("off")

            # Display transformed image (after transformation)
            transformed_slice = transformed_image_tensor[0, :, :, selected_slice].cpu().numpy()  # Channel-first format
            axs[1].imshow(transformed_slice, cmap="gray")
            axs[1].set_title("Transformed Image (After Transform)")
            axs[1].axis("off")

            plt.show()

            # Increment count of displayed images
            displayed_images += 1

if __name__ == '__main__':
    # Initialize the train and test loaders
    # # Example usage
    train_loader, test_loader = get_data_loaders(
        root_dir='/home/nikhil/g17p3/g17-p3/NBIA-Download',
        excel_path='/home/nikhil/g17p3/g17-p3/Dataset/crlm.xlsx',
        batch_size=2,
        test_size=0.25
    )
    
    # Visualize a few images from train_loader before and after transformations
    print("Visualizing random slices from train_loader before and after transformations:")
    visualize_random_slices_before_after(train_loader, num_images=3)
```
